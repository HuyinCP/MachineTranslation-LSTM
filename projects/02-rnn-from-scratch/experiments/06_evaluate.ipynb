# Greedy translation and BLEU

The decoder starts at `<bos>` and reuses its own prediction until `<eos>` or the length cap. Inspect decoded dev sentences first. SacreBLEU uses `BLEU(tokenize='none')` on both hypotheses and references reconstructed from the saved Vietnamese token IDs. The score is comparable only within this preprocessing pipeline. Evaluate test only after dev-based checkpoint selection.

```python
import sys
from pathlib import Path
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'experiments' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))
import torch
from src.data import load_corpus
from src.inference import load_model, translate_sentence
from src.evaluate import evaluate_bleu
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
corpus = load_corpus(PROJECT_ROOT / 'data')
model = load_model(PROJECT_ROOT / 'checkpoints' / 'best.pt', device)
print(translate_sentence(model, 'How are you?', corpus.en_vocab, corpus.vi_vocab, device))
```

```python
score, signature, examples = evaluate_bleu(model, corpus.dev_pairs, corpus.vi_vocab, device)
print(f'Dev BLEU: {score.score:.2f}; {signature}')
for prediction, reference in examples[:3]:
    print('Prediction:', prediction, '\nReference: ', reference)
```
